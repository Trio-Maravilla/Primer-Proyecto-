# __Filtros iniciales__:
## Paso 2: El Dataset debe filtrarse por edad y de preferencia se deben eliminar las columnas innecesarias para evitar confusiones.

### Como se puede ver, el dataset contaba con un montón de columnas que nos resultaron innecesarias para realizar los análisis en base a nuestro objetivo. Por esta razón cihas columnas fueron eliminadas de forma total y ahora sí se puede trabajar con datos que resulten relevantes para nuestro análisis. 

```python
# CELDA 3:
# Primer filtro de edad​
df_filtrado = df[(df["P6040"] >= 18)]

#Eliminación de las columnas innecesarias.
df_filtrado = df_filtrado.drop(["DIRECTORIO","SECUENCIA_ENCUESTA","SECUENCIA_P", "ORDEN", "FEX_C", "P6051", "P1894", "P3510S1", "P3510S1A1", "P3510S1A2", "P3510S1A3", "P3510S1A4", "P3510S1A5", "P3510S1A6", "P3510S1A7",
                  "P3510S1A8", "P3510S1A9", "P3510S1A10", "P3510S1A11", "P3510S2", "P3510S2A1", "P3510S2A2", "P3510S2A4", "P3510S2A5", "P3510S2A6", "P3510S2A7",
                  "P3510S2A8", "P3510S2A9", "P3510S2A10", "P3510S2A11", "P6020", "P6034", "P6071", "P6071S1", "P756", "P756S1", "P756S2", "P756S3", "P3510",
                  "P3510S1", "P3510S2", "P3510S1A1", "P3510S1A2", "P3510S1A3", "P3510S1A4", "P3510S1A5", "P3510S1A6", "P3510S1A7", "P3510S1A8","P3510S1A9", 
                  "P3510S1A10", "P3510S1A11", "P3510S2A1", "P3510S2A2", "P3510S2A3", "P3510S2A4", "P3510S2A5", "P3510S2A6", "P3510S2A7", "P3510S2A8", "P3510S2A9", 
                  "P3510S2A10", "P3510S2A11", "P6074", "P755", "P755S1", "P755S2","P755S3", "P754", "P753", "P753S1", "P753S2", "P752", "P1662", "P6081", "P6081S1", 
                  "P6087", "P6083", "P6083S1", "P6088", "P2057", "P2059", "P1901", "P2061", "P1903", "P1904", "P3038", "P753S3"], axis=1)

# Renombrado de las variables para mejorar su legibilidad.
df_filtrado.columns = ["Jefe_de_hogar","Edad","Estado_civil",
              "Autorreconocimiento_etnico", "Satisfaccion_vida", "Satisfaccion_ingreso", "Satisfaccion_salud", "Satisfaccion_seguridad", "Satisfaccion_trabajo", "Satisfaccion_tiempo_libre",
             "Sentido_proposito", "Escalera_cantril", "Autorreconocimiento_sexual"
              ]

# Reemplazos de datos en columnas "Autorreconocimiento_etnico" y "Autorreconocimiento_sexual"
reemplazos = {1: "Indigena", 2: "Gitano Rom", 3:"Raizal", 4: "Palenquero", 5: "Afrocombino", 6: "Ningún grupo"} # Remplazos para normalizar el autorreconocimiento etnico
reemplazos_sexo = {1: "Hombre", 2: "Mujer", 3:"Hombre trans", 4: "Mujer trans", 5: "Otro", 9: "No responde"} # Remplazos para normalizar el autorreconocimiento sexual
df_filtrado["Autorreconocimiento_etnico"] = df_filtrado["Autorreconocimiento_etnico"].replace(reemplazos) # .replace para aplicar los cambios 
df_filtrado["Autorreconocimiento_sexual"] = df_filtrado["Autorreconocimiento_sexual"].replace(reemplazos_sexo) # .replace para aplicar los cambios 

# Normalización de valores en columna "Jefe_de_hogar"
df_filtrado["Jefe_de_hogar"] = np.where(df_filtrado["Jefe_de_hogar"] == 1, 1, 2) # Uso np.where() para colocar el condicional que normaliza los valores de la columan Jefe_de_hogar


# Renombramiento de datos en columna "Estado_civil"
condiciones_estado_civil = [
    df_filtrado["Estado_civil"] == 1,
    df_filtrado["Estado_civil"] == 2,
    df_filtrado["Estado_civil"] == 3,
    df_filtrado["Estado_civil"] == 4,
    df_filtrado["Estado_civil"] == 5,
    df_filtrado["Estado_civil"] == 6,
                            ] # Define las condiciones que se deben cumplir en orden para renombrar los datos 
reemplazos_estado_civil = ["Union libre hace menos de 2 años", "Union libre hace 2 años o más", "Viudo/a", "Separado/Divorciado", "Soltero/a", "Casado/a"] # Especifica los remplazos para cada condicional en orden
df_filtrado["Estado_civil"] = np.select(condiciones_estado_civil, reemplazos_estado_civil, default="No especificado") # Uso np.select() para aplicar los condicionales con sus respectivos reemplazos automáticamente

# Crear el nuevo dataset filtrado en la ruta definida.
df_filtrado.to_csv(carpeta_salida_csv / "Caracteristicas y composicion del hogar - Filtrado.csv", index=False)
df_filtrado.head(10)
```

```python
# CELDA 4: 
# Búsqueda de Outliers para la variable de "Edad"
df_filtrado.sort_values("Edad", ascending=False)
Q1 = df_filtrado["Edad"].quantile(0.25)
Q3 = df_filtrado["Edad"].quantile(0.75)
IQR = Q3 - Q1

lim_inferior = Q1 - 1.5 * IQR
lim_superior = Q3 + 1.5 * IQR

OutliersEdad = df_filtrado[(df_filtrado["Edad"] < lim_inferior) | (df_filtrado["Edad"] > lim_superior)]
plt.title("Outliers de Edad", fontsize=16, fontweight="bold")
df_filtrado.boxplot(column="Edad")
plt.savefig(carpeta_salida_figuras / "outliers_edad.png", bbox_inches="tight", dpi=300)
plt.show()
```
